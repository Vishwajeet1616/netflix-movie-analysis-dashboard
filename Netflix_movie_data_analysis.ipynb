```python
#categorizing the vote Averge column
def categorized_column(df,col,labels):
    edges = [df[col].min(),
            df[col].quantile(0.25),
            df[col].quantile(0.50),
            df[col].quantile(0.75), 
            df[col].max()]

    df['Vote_Category'] = pd.cut(df[col], bins=edges , labels= labels,duplicates='drop')
    return df
    
```

```python
labels=['not_popular','below_average','average','popular']

categorized_column(clean_df,'Vote_Average',labels)
```
